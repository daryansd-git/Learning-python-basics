## 1. Import libraries

Matplotlib offers detailed control over figures. Seaborn builds on Matplotlib and provides convenient statistical plots.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid")
```

## 2. Matplotlib line plot

Use a line plot to show change across an ordered sequence. Labels and a title make the plot understandable.


```python
days = [1, 2, 3, 4, 5, 6, 7]
temperatures = [18, 20, 19, 23, 25, 24, 27]

plt.figure(figsize=(8, 4))
plt.plot(days, temperatures, marker="o", linewidth=2, label="Temperature")
plt.title("Temperature During One Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.xticks(days)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Matplotlib bar chart

Bar charts compare values across categories. `plt.barh()` creates horizontal bars.


```python
subjects = ["Python", "Math", "Science", "English"]
scores = [88, 76, 92, 81]

plt.figure(figsize=(8, 4))
plt.bar(subjects, scores)
plt.title("Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

## 4. Matplotlib scatter plot

Scatter plots show the relationship between two numeric variables. Association alone does not prove causation.


```python
hours = [1, 2, 2.5, 3, 4, 4.5, 5, 6]
exam_scores = [52, 58, 65, 63, 72, 78, 82, 91]

plt.figure(figsize=(7, 4))
plt.scatter(hours, exam_scores, alpha=0.8)
plt.title("Study Time and Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

## 5. Matplotlib histogram

A histogram shows the distribution of a numeric variable. Try changing `bins` to see how the display changes.


```python
rng = np.random.default_rng(42)
exam_results = rng.normal(loc=75, scale=10, size=300)

plt.figure(figsize=(7, 4))
plt.hist(exam_results, bins=15, edgecolor="black")
plt.title("Distribution of Exam Results")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

## 6. Prepare tabular data for Seaborn

Seaborn works especially well with pandas DataFrames. This dataset is illustrative, not real-world data.


```python
df = pd.DataFrame({
    "study_hours": [1, 2, 2.5, 3, 4, 4.5, 5, 6, 2, 5.5, 3.5, 4.8],
    "score": [52, 58, 65, 63, 72, 78, 82, 91, 60, 86, 70, 80],
    "class": ["A", "A", "B", "A", "B", "B", "A", "B", "B", "A", "A", "B"]
})
df.head()
```

## 7. Seaborn scatter plot

The `hue` argument colors observations by category and automatically adds a legend.


```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="study_hours", y="score", hue="class", s=90)
plt.title("Study Hours vs Score by Class")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

## 8. Seaborn bar plot

`sns.barplot()` shows a statistical estimate (the mean by default) for each category. `errorbar=None` hides uncertainty intervals. Use `sns.countplot()` for counts.


```python
plt.figure(figsize=(7, 4))
sns.barplot(data=df, x="class", y="score", errorbar=None)
plt.title("Average Score by Class")
plt.xlabel("Class")
plt.ylabel("Average Score")
plt.tight_layout()
plt.show()
```

## 9. Seaborn box plot

Box plots summarize the median, quartiles, spread, and potential outliers. They are useful for comparing distributions across groups.


```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="class", y="score")
plt.title("Score Distribution by Class")
plt.xlabel("Class")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

## 10. Seaborn pair plot

Pair plots show pairwise relationships between numeric variables and can color observations by category. They can become cluttered for large datasets.


```python
sns.pairplot(df, hue="class", corner=True)
plt.show()
```

## 11. Seaborn heatmap and correlation

Correlation ranges from -1 to 1 and describes linear association. It does not establish causation and can miss nonlinear patterns.


```python
correlations = df[["study_hours", "score"]].corr()
plt.figure(figsize=(5, 4))
sns.heatmap(correlations, annot=True, cmap="vlag", vmin=-1, vmax=1, fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 12. Save a chart

Save before calling `show()`. The `dpi` parameter controls resolution.


```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="study_hours", y="score", hue="class")
plt.title("Study Hours and Scores")
plt.tight_layout()
plt.savefig("study_hours_vs_scores.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 13. Choosing a chart

| Goal | Chart | Function |
|---|---|---|
| Change over time/order | Line | `plt.plot()` / `sns.lineplot()` |
| Compare categories | Bar | `plt.bar()` / `sns.barplot()` |
| Relationship between two numeric variables | Scatter | `plt.scatter()` / `sns.scatterplot()` |
| Distribution of one numeric variable | Histogram | `plt.hist()` / `sns.histplot()` |
| Compare distributions across groups | Box plot | `sns.boxplot()` |
| Explore pairwise relationships | Pair plot | `sns.pairplot()` |
| Show a matrix | Heatmap | `sns.heatmap()` |

## 14. Good plotting habits
- Add a meaningful title and label both axes, including units where relevant.
- Match the chart type to your question and variable types.
- Keep colors, legends, and annotations purposeful.
- Avoid misleading axis limits, especially for bar charts.
- Check missing values, data types, and outliers before interpretation.
