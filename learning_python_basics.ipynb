## 2. Variables and basic data types

Variables store values. Python commonly uses integers, floats, strings, and booleans.

```python
name = "Alex"       # string (text)
age = 20             # integer (whole number)
height = 1.75        # float (decimal number)
is_learning = True   # boolean (True or False)

print(name, age, height, is_learning)
print(type(name), type(age), type(height), type(is_learning))
```

## 4. Operators

```python
a = 10
b = 3
print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # division
print(a // b)  # floor division
print(a % b)   # remainder
print(a ** b)  # exponentiation
print(a > b)   # comparison
```

## 5. Strings and f-strings

Strings are text. f-strings make it easy to insert values into a message.

```python
first_name = "Sam"
language = "Python"
message = f"{first_name} is learning {language}."
print(message)
print(language.lower())
print(language.upper())
print(language[0])
print(len(language))
```

## 6. Lists, tuples, sets, and dictionaries

```python
fruits = ["apple", "banana", "cherry"]  # list: ordered and changeable
fruits.append("orange")
print(fruits)
print(fruits[0])

coordinates = (10, 20)  # tuple: ordered and immutable
unique_numbers = {1, 2, 2, 3}  # set: unique values
person = {"name": "Alex", "age": 20}  # dictionary: key-value pairs
print(coordinates, unique_numbers, person)
print(person["name"])
```

## 7. Conditions

Use `if`, `elif`, and `else` to choose which code runs. Indentation matters.

```python
score = 82
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Keep practicing!")
```

## 8. Loops

```python
for number in range(1, 6):
    print(number)

countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Go!")
```

## 9. Functions

Functions package reusable code. Parameters are inputs; `return` sends a result back.

```python
def greet(name):
    return f"Hello, {name}!"

def add(x, y):
    return x + y

print(greet("Taylor"))
print(add(4, 7))
```

## 10. List comprehensions

A compact way to build lists from an iterable.

```python
squares = [number ** 2 for number in range(1, 6)]
even_numbers = [number for number in range(10) if number % 2 == 0]
print(squares)
print(even_numbers)
```

## 11. Handling errors

Use `try` and `except` to handle expected problems gracefully.

```python
text_number = "123"
try:
    number = int(text_number)
    print(number * 2)
except ValueError:
    print("That text cannot be converted to an integer.")
```

## 12. Mini project: a simple quiz

Change the questions or add more.

```python
questions = [
    ("What keyword defines a function?", "def"),
    ("Which type stores key-value pairs?", "dictionary"),
    ("What function displays output?", "print"),
]

score = 0
for question, answer in questions:
    print(question)
    # Replace the next line with input(question) to make it interactive.
    user_answer = answer  # Demo mode: correct answers are pre-filled.
    if user_answer.strip().lower() == answer:
        score += 1
print(f"You scored {score} out of {len(questions)}!")
```
